Importing libraries:

```python
import pickle
import pandas as pd
```

Load Dataset and Model:

```python
dataset=pd.read_csv("CKD_Dataset.csv")
loaded_model=pickle.load(open('finalized_model_Random_forest.sav','rb'))
```

Input/Output split:

```python
independent=dataset.drop('classification',axis=1)
```

```python
dependent=dataset['classification']
```

Convert Nominal to Numerical datas:

```python
independent=pd.get_dummies(independent,drop_first=True,dtype=int)
```

Take Random row as an Input from the Dataset

```python
new_data=independent.sample(n=1)
```

Matching Training and Testing Columns

```python
expected_columns = loaded_model.feature_names_in_
new_data = new_data.reindex(
    columns=expected_columns,
    fill_value=0
)
```

Final Prediction:

```python
result=loaded_model.predict(new_data)
print(f"Chronic Kidney Disease Prediction : {result[0]}")
```
